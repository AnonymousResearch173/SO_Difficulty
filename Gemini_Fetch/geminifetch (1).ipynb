```python
import pandas as pd
import os

# Function to split CSV into multiple files of n rows each
def split_csv(input_file, output_dir, rows_per_file=15):
    # Read the input CSV file
    df = pd.read_csv(input_file)
    
    # Calculate the number of split files needed
    num_files = (len(df) // rows_per_file) + 1
    
    # Create the output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    # Split the DataFrame and save each split to a separate CSV file
    for i in range(num_files):
        start_idx = i * rows_per_file
        end_idx = min((i + 1) * rows_per_file, len(df))
        split_df = df.iloc[start_idx:end_idx]
        split_df.to_csv(os.path.join(output_dir, f'split_{i}.csv'), index=False)
split_csv(r"Java_2018.csv",r"Java_split", 15)
```

```python
def process_split_files(input_dir, output_dir):
    # List all files in the input directory
    files = os.listdir(input_dir)
    processed_files = []
    
    # Process each split file
    for file in files:
        if file.endswith('.csv'):
            # Perform your specific operation here
            file_path = os.path.join(input_dir, file)
            print(f"Processing {file_path}:")
            df = pd.read_csv(file_path)
            
            # Placeholder for your actual processing code
            processed_df = process(df)  # Assuming no processing, just renaming
            
            # Save processed DataFrame to a new CSV file
            processed_file_path = os.path.join(output_dir, f'z{file}')
            processed_df.to_csv(processed_file_path, index=False)
            
            # Delete the original split file
            os.remove(file_path)
            
            processed_files.append(processed_df)
    
    return processed_files
process_split_files(r"Java_split", r"Java_output")
```

```python
import pandas as pd
import glob
import os

def combine_csv_files(directory_path):
    # Get a list of all CSV files in the specified directory
    csv_files = glob.glob(os.path.join(directory_path, '*.csv'))

    # Initialize an empty list to store dataframes
    dfs = []

    # Read each CSV file and append it to the list
    for file in csv_files:
        df = pd.read_csv(file)
        dfs.append(df)

    # Concatenate the dataframes row-wise
    combined_df = pd.concat(dfs, axis=0, ignore_index=True)

    # Get the directory name to save the combined CSV file in the same directory
    output_directory = os.path.dirname(directory_path)

    # Save the combined dataframe to a new CSV file in the same directory
    output_file = os.path.join(output_directory, 'All_text_combined_final.csv')
    combined_df.to_csv(output_file, index=False)

    print(f"Combined CSV file saved at: {output_file}")

# Example usage:
directory_path = r"Java_output"
combine_csv_files(directory_path)
```
